```python
import numpy
```

```python
def show_equalities(n, mtrx):
    print("\nDaftar persamaan:")
    for baris in mtrx:
        if abs(baris[0]) != 1:
            print(f'{baris[0]:.0f}', end="")
        if baris[0] == -1:
            print("-", end="")
        print("x1", end="")
        for i in range(1, n):
            if baris[i] == 0:
                continue
            if baris[i] < 0:
                print(" - ", end="")
            else:
                print(" + ", end="")
            if abs(baris[i]) != 1:
                print(f'{abs(baris[i]):.0f}', end="")
            print(f'x{i + 1}', end="")
        print(" = ", end="")
        print(baris[n])
```

```python
def show_matrice_multiplication(n, a, b, c, mnn, mn1):
    print(f'\n\n\t{a}.{b} = {c}\n')
    for i in range(n):
        print("    [\t", end='')
        for j in range(n):
            print(f'{mnn[i][j]:.3f}\t', end='')
        print("]", end='')

        print(f' [ {b}{i+1} ]  =  [ {mn1[i][0]:.3f} ]')
```

```python
def l_u_decomp(n, mtrx):
    u = numpy.zeros((n, n))
    l = numpy.zeros((n, n))
    b = numpy.zeros((n, 1))
    y = numpy.zeros((n, 1))
    x = numpy.zeros((n, 1))

    # "A = L.U" pada iterasi ke-0 adalah  "A = I.A"
    # a. sehingga U di iterasi ke-0 adalah duplikat dari A
    for i in range(n):
        for j in range(n+1):
            if j != n:
                u[i][j] = mtrx[i][j]
            else:
                b[i][0] = mtrx[i][j]
    # b. sehingga L di iterasi ke-0 adalah matriks identitas (I) berukuran n x n
    for i in range(n):
        for j in range(n):
            if i == j:
                l[i][j] = 1
    
    show_matrice_multiplication(n, "A", "x", "b", mtrx, b)

    # iterasi dekomposisi L-U
    for i in range(n):
        for j in range(n-1, i, -1):
            faktor = u[j][i]/u[i][i]
            l[j][i] = faktor
            for k in range(n):
                u[j][k] -= faktor*u[i][k]

    # mencetak "A = L.U" hasil setelah iterasi terakhir
    print("\n\n\tA = L.U\n")
    for i in range(n):
        print("[  ", end='')
        for j in range(n):
            print(f'  {mtrx[i][j]}\t', end='')
        print("]", end='')

        if i == int(n/2):
            print("   =\t", end='')
        else:
            print('\t', end='')

        print("[    ", end='')
        for j in range(n):
            print(f'{l[i][j]:.3f}\t', end='')
        print("]", end='')

        print(" [   ", end='')
        for j in range(n):
            print(f'{u[i][j]:.3f}\t', end='')
        print("]")

    show_matrice_multiplication(n, "L", "y", "b", l, b)

    # mencari nilai-nilai y
    for i in range(n):
        subst = 0
        for j in range(0, i, 1):
            subst += l[i][j]*y[j][0]
        y[i][0] = b[i][0] - subst

    print("\nDidapatkan y:\n")
    for i in range(n):
        print(f'    [   {y[i][0]:.3f}\t]')

    show_matrice_multiplication(n, "U", "x", "y", u, y)

    # mencari nilai-nilai x
    for i in range(n-1, -1, -1):
        subst = 0
        for j in range(n-1, i, -1):
            subst += u[i][j]*x[j][0]
        x[i][0] = (y[i][0] - subst)/u[i][i]
        
    print("\nDidapatkan x:\n")
    for i in range(n):
        print(f'    [   {x[i][0]:.3f}\t]')
```
