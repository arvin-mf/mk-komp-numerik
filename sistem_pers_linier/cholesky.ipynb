```python
def transpose(n, mtrx):
    newMtrx = [row[:] for row in mtrx]
    for i in range(n):
        for j in range(i, n):
            temp = newMtrx[i][j]
            newMtrx[i][j] = newMtrx[j][i]
            newMtrx[j][i] = temp
    return newMtrx
```

```python
def cholesky(n, mtrx):
    b = numpy.zeros((n, 1))
    for i in range(n):
        b[i][0] = mtrx[i][n]

    showMatriceMultiplication(n, "A", "x", "b", mtrx, b)

    u = numpy.zeros((n, n))
    for i in range(n):
        subst = 0
        for j in range(i):
            subst += numpy.square(u[j][i])
        u[i][i] = numpy.sqrt(mtrx[i][i] - subst)
        for j in range(i+1, n):
            subst = 0
            for k in range(i):
                subst += u[k][i]*u[k][j]
            u[i][j] = (mtrx[i][j] - subst)/u[i][i]
    
    # menyalin matriks U
    l = numpy.zeros((n, n))
    for i in range(n):
        for j in range(n):
            l[i][j] = u[i][j]
    l = transpose(n, l)

    showAAsLUProduct(n, mtrx, l, u)

    showMatriceMultiplication(n, "L", "y", "b", l, b)

    y = numpy.zeros((n, 1))
    # mencari nilai-nilai y
    for i in range(n):
        subst = 0
        for j in range(0, i, 1):
            subst += l[i][j]*y[j][0]
        y[i][0] = (b[i][0] - subst)/l[i][i]

    print("\nDidapatkan y:\n")
    for i in range(n):
        print(f'    [   {y[i][0]:.3f}\t]')

    showMatriceMultiplication(n, "U", "x", "y", u, y)

    x = numpy.zeros((n, 1))
    # mencari nilai-nilai x
    for i in range(n-1, -1, -1):
        subst = 0
        for j in range(n-1, i, -1):
            subst += u[i][j]*x[j][0]
        x[i][0] = (y[i][0] - subst)/u[i][i]
        
    print("\nDidapatkan x:\n")
    for i in range(n):
        print(f'    [   {x[i][0]:.3f}\t]')
    
```

```python
seederNum = 3
seederEqu = [
    [2, 3, -1, -2],
    [3, 10, 1, -1],
    [-1, 1, 2, 3]
]
showEqualities(seederNum, seederEqu)

cholesky(seederNum, seederEqu)
```
